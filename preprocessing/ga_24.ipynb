```python
import geopandas
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../data/Georgia/ga24.csv")
```

```python
print(f"shape: {df.shape}")
df.head()
```

```python
cols = df.columns
print(f"all columns: {cols}")
```

```python
df_sorted = df.sort_values(by=['county_fips', 'precinct'])
df_sorted.head()
```

```python
df_sorted['GEOID'] = df_sorted['county_fips'].astype(str) + "-" + df_sorted['precinct'].astype(str)
```

```python
df_sorted.head()
```

```python
column_to_drop = []
for col in cols:
    unique = df_sorted[col].unique()
    if len(unique)==1:
        print(f"{col} has only 1 unique value: {unique}")
        column_to_drop.append(col)
column_to_drop.append('precinct')
column_to_drop.append('county_fips')
column_to_drop
```

```python
df_sorted = df_sorted.drop(column_to_drop, axis =1)
```

```python
cols = df_sorted.columns.tolist()
cols.remove('GEOID')
cols = ['GEOID'] + cols

df_sorted = df_sorted[cols]
df_sorted
```

```python
candidates = df_sorted['candidate'].unique()
print(f"total number of candidates: {len(candidates)}")
print(f"candidates:\n {candidates}")
```

```python
first = df_sorted[df_sorted['GEOID'] == '10001-11-05']
print(f"num of first precinct: {first.shape}")
print(f"columns:{first.columns}")
```
