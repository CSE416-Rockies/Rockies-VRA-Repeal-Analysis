```python
import pandas as pd
import numpy as np
import geopandas
```

```python
df_de = geopandas.read_file("../data/Delaware/Precinct/de_nyt.geojson")
df_de = df_de.sort_values(by="GEOID")

columns_needed = ['GEOID', 'geometry']
de_precincts = df_de[columns_needed]
de_precincts['precicnt'] = "Precicnt " + df_de["GEOID"]
```

```python
de_precincts
```

```python
df_ga = geopandas.read_file("../data/Georgia/Precinct/ga_nyt.geojson")
ga_precincts = df_ga[columns_needed]
ga_precincts['precicnt'] = "Precicnt " + df_ga["GEOID"]
```

```python
ga_precincts
```

```python
print(ga_precincts['precicnt'].isnull().sum())
print(de_precincts['precicnt'].isnull().sum())
```

```python
de_precincts.to_file("Delaware_precincts.geojson", driver="GeoJSON")
ga_precincts.to_file("Georgia_precincts.geojson", driver="GeoJSON")
```
