# Arkansas VTD (Voting Tabulation District) Shape File
##### *Census approximations of precincts.*

The Arkansas GIS Office is pleased to announce we have added two new fields to our Elections Precincts data. 
1) The new VTDST fields correlate with the 2010 and 2020 Census VTD file that is included in the PL 94-171 Redistricting Data Summary Files. 
2) The VTDST can be considered the Census code for a particular Election Precinct and is unique per county only. 
aka: vtdst20 – this is the Census VTDST place for the 2020 Census redistricting files. 


GEOID20 = STATEFP + COUNTYFP + VTDST

https://gis.arkansas.gov/2021/08/18/new-fields-added-to-election-precincts-data
