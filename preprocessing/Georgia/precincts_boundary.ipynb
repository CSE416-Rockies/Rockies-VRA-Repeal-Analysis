```python
import pandas as pd
import numpy as np
import geopandas
```

```python
columns_needed = ['GEOID', 'geometry']
df_ga = geopandas.read_file("../data/Georgia/Precinct/ga_nyt.geojson")
ga_precincts = df_ga[columns_needed]
ga_precincts['precinct'] = "Precinct " + df_ga["GEOID"]
```

```python
print(ga_precincts.shape)
ga_rows = len(ga_precincts)
print(ga_rows)
```

```python
ga_precincts.loc[:, "total_population"] = np.random.randint(1000, 5001, size=ga_rows)
ga_precincts.loc[:, "latino_percentage"] = np.random.uniform(0,80, size=ga_rows)
ga_precincts.loc[:, "black_percentage"] = (100-ga_precincts['latino_percentage'])*np.random.uniform(0, 1, size=ga_rows)
ga_precincts.loc[:, "other_percentage"] = (100 - ga_precincts['latino_percentage'] - ga_precincts['black_percentage'])
```

```python
print(ga_precincts['precinct'].isnull().sum())
```

```python
ga_precincts.to_file("Georgia_precincts.geojson", driver="GeoJSON")
```
