## Georgia Box Whisker Plot Data

### Imports

```python
import pandas as pd
import json
```

### District Data Path to DataFrame

```python
path = '../output/Arkansas/ar-district-CVAP.csv'
```

```python
district_cvap = pd.read_csv(path)
district_cvap
```

```python
district_cvap.drop(columns=['Unnamed: 0', 'State'], inplace=True)
district_cvap
```

### Race CVAP Enacted Plan Points

```python
dfs = [None, None, None, None]
groups = ['White', 'Black', 'Latino', 'Other']
```

```python
def calculate_enacted_points(dfs, groups):
    for i, group in enumerate(groups):
        df = district_cvap[['Total CVAP', f'{group} CVAP']]
        df = df.sort_values(by=f'{group} CVAP').reset_index().drop(columns='index').reset_index()
        df['share'] = df[f'{group} CVAP'] / df['Total CVAP']
        df = df[['share']]

        dfs[i] = df
 
calculate_enacted_points(dfs, groups)

for df in dfs:
    print(df)
```

```python
white_share = dfs[0].to_dict()
black_share = dfs[1].to_dict()
latino_share = dfs[2].to_dict()
other_share = dfs[3].to_dict()

ar_enacted_shares = {'white': white_share, 'black': black_share, 'latino': latino_share, 'other': other_share}
```

```python
with open('../output/Arkansas/ar-enacted-shares.json','a') as file:
    file.write(json.dumps(ar_enacted_shares,indent=4))
```
