```python
import geopandas
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../data/Delaware/de24_cleaned.csv")
```

```python
print(f"shape: {df.shape}")
df.head()
```

```python
cols = df.columns
print(f"all columns: {cols}")
```

```python
df_sorted = df.sort_values(by=['county_fips', 'precinct'])
df_sorted.head()
```

```python
df_sorted['GEOID'] = df_sorted['county_fips'].astype(str) + "-" + df_sorted['precinct'].astype(str)
```

```python
df_sorted.head()
```

```python
column_to_drop = []
for col in cols:
    unique = df_sorted[col].unique()
    if len(unique)==1:
        print(f"{col} has only 1 unique value: {unique}")
        column_to_drop.append(col)
column_to_drop.append('precinct')
column_to_drop.append('county_fips')
column_to_drop
```

```python
df_sorted = df_sorted.drop(column_to_drop, axis =1)
```

```python
cols = df_sorted.columns.tolist()
cols.remove('GEOID')
cols = ['GEOID'] + cols

df_sorted = df_sorted[cols]
df_sorted
```

```python
df_president = df_sorted[df_sorted['office']=='US PRESIDENT']
```

```python
print(f"df president shape: {df_president.shape}")
candidates = df_president['candidate'].unique()
print(f"total number of candidates: {len(candidates)}")
print(f"unique candiates: {candidates}")
```

```python
df_president = df_president.drop('office', axis = 1)
```

```python
party_detailed = df_president['party_detailed'].unique()
print(f"all parties: {len(party_detailed)}")
print(party_detailed)

party_simplified = df_president['party_simplified'].unique()
print(f"all simplified parties: {len(party_simplified)}")
print(party_simplified)
```

```python
print(f"unique county: {df_president['county_name'].unique()}")
print(f"unique jurisdiction_name: {df_president['jurisdiction_name'].unique()}")
print(f"unique jurisdictino_fips: {df_president['jurisdiction_fips'].unique()}")
print(f"unique district: {df_president['district'].unique()}")
print(f"unique dataverse: {df_president['dataverse'].unique()}")
```

```python
df_president = df_president.drop('district', axis=1)
df_president = df_president.drop('dataverse', axis = 1)
```

```python
df_president.head()
```

```python
first_precicnt = df_president[df_president['GEOID']=='10001-11-05']
print(f"total first precicnt: {first_precicnt.shape}")
first_precicnt.head()
```

```python
for col in df_president.columns:
    if df_president[col].isna().any():
        print(f"NaN found in {col}")
```

importing NYT DE data to get the geometry boundry values

```python
df_geometry = geopandas.read_file("../data/Delaware/Precinct/de_nyt.geojson")
```

```python
df_geometry.columns
```

```python
needed_cols = ['GEOID', 'geometry']
df_geometry = df_geometry[needed_cols]
df_geometry
```

```python
len(df_geometry['GEOID'].unique())
```

merging by **GEOID**

```python
df_merged = df_president.merge(
    df_geometry[['GEOID', 'geometry']],
    on='GEOID',
    how='left'
)
```

```python
total_unique = df_merged['GEOID'].unique()
print(f"total num of precinct: {len(total_unique)}")
```
