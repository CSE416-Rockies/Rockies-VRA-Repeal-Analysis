## Rough Proportion

```python
MINORITIES = ['Black','Latino', 'Other']
```

```python
def calc_rough(proportions, effective, state='AR'):
    rough = {}
    total_district = 4 if state=='AR' else 14
    for minority in MINORITIES:
        num_effective = effective.loc[effective["Demographic"] == minority, "Effective_Districts"].iloc[0]
        proportion = proportions.loc['state_wide'][minority]
        rough[minority] = (num_effective/total_district)/proportion
    
    return rough
```

### Arkansas

```python
ar_proportion = pd.read_json('output/Arkansas/ar_cvap_proportion.json')
ar_effective = pd.read_json('output/Arkansas/ar_enacted_effective.json')
```

```python
ar_proportion = pd.DataFrame.from_dict(data=ar_proportion, orient='columns')
ar_effective = pd.DataFrame.from_dict(data=ar_effective, orient='columns')
```

```python
ar_rough_proportion = calc_rough(ar_proportion, ar_effective, state='AR')
```

```python
ar_rough_proportion
```

### Georgia

```python
ga_proportion = pd.read_json('output/Georgia/ga_cvap_proportion.json')
ga_effective = pd.read_json('output/Georgia/ga_enacted_effective.json')
ga_proportion = pd.DataFrame.from_dict(data=ga_proportion, orient='columns')
ga_effective = pd.DataFrame.from_dict(data=ga_effective, orient='columns')
```

```python
ga_rough_proportion = calc_rough(ga_proportion, ga_effective, state='GA')
```

```python
ga_rough_proportion
```

### Combine

```python
rough_proportions = {}
rough_proportions['Arkansas'] = ar_rough_proportion
rough_proportions['Georgia'] = ga_rough_proportion
```

```python
rough_proportions
```

```python
rough_df = pd.DataFrame.from_dict(rough_proportions, orient='columns')
```
