```python
import pandas as pd
import numpy as np
```

```python
df_de = pd.read_csv("../data/Delaware/DE_VTD_Hispanic.csv")
print(f"DE shape: {df_de.shape}")
df_de.head(5)
```

```python
df_de = df_de.T
# 1. Turn the 'Label (Grouping)' index into a regular column
df_de = df_de.reset_index()

# 2. Set the columns to the first row's values
df_de.columns = df_de.iloc[0]

# 3. Drop that first row (which is now redundant)
df_de = df_de[1:]

# 4. Clean up the index so it starts at 0 again
df_de = df_de.reset_index(drop=True)

# 5. Remove the '0' label that often sits above the index
df_de.columns.name = None

print(df_de.shape)
```

```python
df_de.head(5)
```

```python
df_de.columns = df_de.columns.str.strip()
columns = df_de.columns
print(columns)
```

```python
latino_related = []
black_related = []
alone = []
for col in columns:
    if "latino" in col or "hispanic" in col.lower():
        latino_related.append(col)
    if "black" in col.lower():
        black_related.append(col)
    if "alone" in col.lower():
        alone.append(col)

print(f"latino related: {latino_related}")
print(f"black related: {black_related}")
print(f"alone: {alone}")
```

```python
columns_needed = ['Label (Grouping)', 'Hispanic or Latino'] + alone
columns_needed
```

```python
df_de = df_de[columns_needed]
df_de
```

```python
cols_to_sum = [
    'White alone', 
    'American Indian and Alaska Native alone', 
    'Asian alone', 
    'Native Hawaiian and Other Pacific Islander alone',
    'Some Other Race alone'
]

for col in cols_to_sum:
    df_de[col] = pd.to_numeric(df_de[col].astype(str).str.replace(',', ''), errors='coerce').fillna(0)

df_de['Others'] = df_de[cols_to_sum].sum(axis=1)
df_de
```

```python
df_de = df_de.drop(cols_to_sum, axis=1)
df_de.head(5)
```

```python
df_ga = pd.read_csv("../data/Georgia/GA_VTD_Hispanic.csv")
print(f"GA shape: {df_ga.shape}")
df_ga.head(5)
```

```python
df_ga.columns = df_ga.iloc[0]
df_ga = df_ga[1:]
df_ga = df_ga.reset_index(drop=True)
df_ga = df_ga.rename(columns={'Geography':'GEOID'})
df_ga.head(5)
```

```python
df_ga.columns = df_ga.columns.astype(str).str.strip()
columns = df_ga.columns
alone = []
for col in columns:
    
    if "alone" in col.lower():
        alone.append(col)
        
print(f"all alone: {alone}")

alone.remove('!!Total:!!Not Hispanic or Latino:!!Population of one race:!!Black or African American alone')
print(f"after: {alone}")
```

```python
columns_needed = (['GEOID', '!!Total:!!Hispanic or Latino', 
                   '!!Total:!!Not Hispanic or Latino:!!Population of one race:!!Black or African American alone']
                  + alone)

df_ga = df_ga[columns_needed]
df_ga.head(5)
```

```python
for col in alone:
    df_ga[col] = pd.to_numeric(df_ga[col].astype(str).str.replace(',', ''), errors='coerce').fillna(0)

df_ga['Others'] = df_ga[alone].sum(axis=1)
df_ga = df_ga.drop(alone, axis=1)
df_ga.head(5)
```

```python
df_ga.columns
```

```python
change = {'!!Total:!!Hispanic or Latino':'Hispanic or Latino', 
          '!!Total:!!Not Hispanic or Latino:!!Population of one race:!!Black or African American alone': 'Black or African American alone'}
df_ga = df_ga.rename(columns=change)
df_ga.head(5)
```

```python
df_ga.shape
```

```python
starts_with_total = df_ga['GEOID'].str.startswith('7000000US')
if False in starts_with_total:
    print("there are id doesn't start with 7000000US")
```

```python
df_ga['GEOID'] = df_ga['GEOID'].str.replace('7000000US','',regex=False)
df_ga.head(5)
```

```python
df_ga['GEOID'] = df_ga['GEOID'].astype(str)
df_ga['GEOID'] = df_ga['GEOID'].str[:5] + '-' + df_ga['GEOID'].str[5:]
df_ga.head(5)
```

```python
df_ga = df_ga.sort_values(by='GEOID')
df_ga.head(5)
```

```python
df_precincts = pd.read_csv('GA_precincts.csv')
df_precincts = df_precincts.drop('Unnamed: 0', axis=1)
df_precincts = df_precincts.sort_values(by='GEOID')
df_precincts.head(5)
```

```python
# temp_df = df_precincts['GEOID'].str.split('-', expand=True)
# temp_df[1] = temp_df[1].str.zfill(6)
# df_precincts['GEOID'] = temp_df[0] + '-' + temp_df[1]

```

```python
df_precincts = df_precincts.sort_values(by='GEOID')
df_precincts
```

```python
def clean_to_bones(geoid):
    # 1. Convert to string and uppercase
    geoid = str(geoid).upper()
    # 2. Split by the hyphen
    if '-' in geoid:
        county, vtd = geoid.split('-',1)
    else:
        # If no hyphen, assume first 5 are county (standard for GA)
        county, vtd = geoid[:5], geoid[5:]
    
    # 3. Strip all leading zeros from the VTD part
    vtd_clean = vtd.lstrip('0')
    
    return f"{county}-{vtd_clean}"

# Apply this to BOTH dataframes
df_ga['bone_id'] = df_ga['GEOID'].apply(clean_to_bones)
df_precincts['bone_id'] = df_precincts['GEOID'].apply(clean_to_bones)
```

```python
num_duplicates = df_ga['bone_id'].duplicated().sum()
print(f"There are {num_duplicates} duplicate IDs in df_ga.")

num_duplicates = df_precincts['bone_id'].duplicated().sum()
print(f"There are {num_duplicates} duplicate IDs in df_precincts.")
```
