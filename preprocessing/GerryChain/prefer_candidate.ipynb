```python
import numpy as np
import pandas as pd
import geopandas as gpd
```

```python
ga_df = gpd.read_file("../output/Georgia/ga_seawulf.gpkg")
ar_df = gpd.read_file("../output/Arkansas/ar_seawulf.gpkg")
```

```python
ga_df.columns
```

```python
races = ['White', 'Black', 'Latino', 'Other']

for race in races:
    ar_df[f"{race}_harris_votes"] = ar_df[f"{race}_population"]*ar_df[f"{race}_harris_pct"]
    ar_df[f"{race}_trump_votes"] = ar_df[f"{race}_population"]*ar_df[f"{race}_trump_pct"]
    ga_df[f"{race}_harris_votes"] = ga_df[f"{race}_population"]*ga_df[f"{race}_harris_pct"]
    ga_df[f"{race}_trump_votes"] = ga_df[f"{race}_population"]*ga_df[f"{race}_trump_pct"]
    
```

```python
columns_drop = ['Kamala d. harris', 'Donald j. trump',
       'Other_candidates', 'Total_votes', 'White_population',
       'Black_population', 'Latino_population', 'Other_population',
       'Total_population', 'White_harris_pct', 'Black_harris_pct',
       'Latino_harris_pct', 'Other_harris_pct', 'White_trump_pct',
       'Black_trump_pct', 'Latino_trump_pct', 'Other_trump_pct',]
```

```python
ar_prefer_df = ar_df.drop(columns=columns_drop)
ga_prefer_df = ga_df.drop(columns=columns_drop)
```

```python
ar_grouped = ar_prefer_df.groupby("District").sum(numeric_only=True)
```

```python
ar_grouped
```

```python
ar_preferred_candidates = {}
for district_id, district in ar_grouped.iterrows():
    
    ar_preferred_candidates[district_id] = {}
    for race in races:
        cand = "Trump"
        if district[f"{race}_harris_votes"] > district[f"{race}_trump_votes"]:
            cand = "Harris"
        ar_preferred_candidates[district_id][race] = cand

ar_preferred_candidates
```

```python
ga_grouped = ga_prefer_df.groupby("District").sum(numeric_only=True)
```

```python
ga_grouped
```

```python
ga_preferred_candidates = {}
for district_id, district in ga_grouped.iterrows():
    
    ga_preferred_candidates[district_id] = {}
    for race in races:
        cand = "Trump"
        if district[f"{race}_harris_votes"] > district[f"{race}_trump_votes"]:
            cand = "Harris"
        ga_preferred_candidates[district_id][race] = cand

ga_preferred_candidates
```

```python
ar_preferred_district = pd.DataFrame.from_dict(ar_preferred_candidates, orient='index')
ga_preferred_district = pd.DataFrame.from_dict(ga_preferred_candidates, orient='index')
```

```python
ar_preferred_district = ar_preferred_district.reset_index()
ar_preferred_district = ar_preferred_district.rename(columns={'index': 'District'})
ar_preferred_district.head()
```

```python
ga_preferred_district = ga_preferred_district.reset_index()
ga_preferred_district = ga_preferred_district.rename(columns={'index': 'District'})
ga_preferred_district.head()
```

```python
ga_preferred_district.to_csv('../output/Georgia/ga_preferred_candidates.csv', index=False)
```

```python
ar_preferred_district.to_csv('../output/Arkansas/ar_preferred_candidates.csv', index=False)
```

```python
ar_state = ar_grouped.sum()
ar_statewide_winner = {}

for race in races:
    h = ar_state[f"{race}_harris_votes"]
    t = ar_state[f"{race}_trump_votes"]
    
    ar_statewide_winner[race] = "Harris" if h > t else "Trump"
    
ar_statewide_winner
```

```python
ga_state = ar_grouped.sum()
ga_statewide_winner = {}

for race in races:
    h = ga_state[f"{race}_harris_votes"]
    t = ga_state[f"{race}_trump_votes"]
    
    ga_statewide_winner[race] = "Harris" if h > t else "Trump"
    
ga_statewide_winner
```

```python
h_total = sum(ga_state[f"{r}_harris_votes"] for r in races)
t_total = sum(ga_state[f"{r}_trump_votes"] for r in races)

ga_statewide_winner_all = "Harris" if h_total > t_total else "Trump"
print(f"overall winner for GA: {ga_statewide_winner_all}")

h_total = sum(ar_state[f"{r}_harris_votes"] for r in races)
t_total = sum(ar_state[f"{r}_trump_votes"] for r in races)

ar_statewide_winner_all = "Harris" if h_total > t_total else "Trump"
print(f"overall winner for AR: {ar_statewide_winner_all}")
    
```

```python
for district_id, district in ar_grouped.iterrows():
    harris = district['Kamala d. harris']
    trump = district['Donald j. trump']
    prefer = ar_preferred_district[ ar_preferred_district['District']== district_id]
    black_preferred = prefer['Black'].iloc[0]
    # print(f"black prefer: {black_preferred}")
    latino_preferred = prefer['Latino'].iloc[0]
    harris_won = False
    if(harris > trump):
        harris_won = True
        print(f"AR: harris won {district_id}")
    
    if black_preferred == "Harris" and harris_won:
        print(f"-- black prefer {black_preferred}, black prefer won")
    elif black_preferred == "Trump" and not harris_won:
        print(f"-- black prefer {black_preferred}, black prefer won")
    
    if latino_preferred == "Harris" and harris_won:
        print(f"-- latino prefer {latino_preferred}, latino prefer won")
    elif latino_preferred == "Trump" and not harris_won:
        print(f"-- latino prefer {latino_preferred}, latino prefer won")

print("*"*50)

ga_black_effective_cnt = 0
ga_latino_effective_cnt = 0
ga_district_scores = {}
for district_id, district in ga_grouped.iterrows():
    ga_district_scores[district_id] = 0
    harris = district['Kamala d. harris']
    trump = district['Donald j. trump']
    # print(f"harris: {harris}, trump: {trump}")
    
    prefer = ga_preferred_district[ ga_preferred_district['District']== district_id]
    black_preferred = prefer['Black'].iloc[0]
    latino_preferred = prefer['Latino'].iloc[0]
    
    harris_won = False
    if(harris  > trump):
        harris_won = True
        print(f"GA: harris won @ district {district_id}") 
    
    minority_won = False
    if black_preferred == "Harris" and harris_won:
        ga_black_effective_cnt +=1
        minority_won = True
        print(f"-- black prefer {black_preferred}, black prefer won")
    elif black_preferred == "Trump" and not harris_won:
        ga_black_effective_cnt +=1
        minority_won = True
        print(f"-- black prefer {black_preferred}, black prefer won")
    
    if latino_preferred == "Harris" and harris_won:
        ga_black_effective_cnt +=1
        minority_won = True
        print(f"-- latino prefer {latino_preferred}, latino prefer won")
    elif latino_preferred == "Trump" and not harris_won:
        ga_black_effective_cnt +=1
        minority_won = True
        print(f"-- latino prefer {latino_preferred}, latino prefer won")
    
    if minority_won:
        ga_district_scores[district_id] = 1

print(ga_district_scores)
print(f"effective black: {ga_black_effective_cnt}, effective latino: {ga_latino_effective_cnt}")
```

group control factor $c = \min(2k, 1)$

```python
ga_black_scores = {}
ga_latino_scores = {}
ga_total_scores = {}
for district_id, score in ga_district_scores.items():
    district_data = ga_df[ga_df['District'] == district_id]
    black_population = district_data['Black_population'].iloc[0]
    latino_population = district_data['Latino_population'].iloc[0]
    total_population = district_data['Total_population'].iloc[0]
    total_min = black_population+latino_population
    print(f"total_population: {total_population}, total_min: {total_min}")
    ga_black_scores[district_id] = min(score, black_population/total_population*2)
    ga_latino_scores[district_id] = min(score, latino_population/total_population*2)
    ga_total_scores[district_id] = min(score, total_min/total_population*2)

print(f"black score: {ga_black_scores}")
print(f"latino score: {ga_latino_scores}")
print(f"total (combine black and latino): {ga_total_scores}")
```

- use threshold 0.6 due to only 1 election available

```python
threshold = 0.6
ga_benchmark_cnt = 0
for district_id, score in ga_black_scores.items():
    if score > 0.6:
        ga_benchmark_cnt+=1

print(f"benchmark effective state: {ga_benchmark_cnt}")
```
